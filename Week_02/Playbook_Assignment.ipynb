# Ticket Tagger

## Step 1 — Understand the Business Problem

### Success Criteria

**Exactly one category, deterministic, strict format.**

### Reflection

Answer briefly before moving on:

1. Should the output always be identical?
   **Yes. When a customer submits a ticket, it needs to get sorted to the exact same department every time.**
2. Is creativity helpful or harmful?
   **Harmful. If the model gets creative, it might invent new category names that our system can't read.**
3. Should the model take risks?
   **No. It should stick strictly to the category list we give it.**

---

## Step 2 — Design Three Configurations

You are **not** searching for the correct answer.

Instead, design:

- **Primary** → Your recommendation
- **Challenger A** → A reasonable alternative
- **Challenger B** → Another reasonable alternative

|Configuration|Temperature|Top-k|Top-p|Penalty|Greedy/Sampling|Why?|
|---|---:|---:|---:|---|---|---|
|Primary|0.0|N/A|1.0|0.0|Greedy|I want the exact same category every time without any random variation.|
|Challenger A|0.2|40|0.9|0.0|Sampling|Testing a very low temperature to see if a tiny bit of randomness breaks the format.|
|Challenger B|0.9|50|0.95|0.0|Sampling|High temperature to test how much randomness ruins strict category output.|

---

## Step 3 — Create One Representative Prompt

Choose ONE prompt that realistically represents this product.

```text
Ticket: "I forgot my password and cannot log in."
Category: Account Access

Ticket: "My screen turns black when I open the app."
Category: Technical Support

Ticket: "I noticed two identical charges of $49.99 on my credit card statement from yesterday. Please help resolve this!"
Category:

````python
# ================================
# Run your experiments here
# ================================

import warnings
warnings.filterwarnings("ignore")

from transformers import pipeline

generator = pipeline("text-generation", model="gpt2")

# Few-shot prompt so GPT-2 understands the completion pattern
prompt = """Ticket: "I forgot my password and cannot log in."
Category: Account Access

Ticket: "My screen turns black when I open the app."
Category: Technical Support

Ticket: "I noticed two identical charges of $49.99 on my credit card statement from yesterday. Please help resolve this!"
Category:"""

# Primary (Greedy)
print("=== Primary (Greedy) ===")
for i in range(3):
    out = generator(prompt, do_sample=False, max_new_tokens=5, pad_token_id=50256)
    result = out[0]['generated_text'][len(prompt):].strip().split('\n')[0]
    print(f"Run {i+1}: {result}")

# Challenger A (Temp 0.2)
print("\n=== Challenger A (Temp 0.2) ===")
for i in range(3):
    out = generator(prompt, do_sample=True, temperature=0.2, top_p=0.9, top_k=40, max_new_tokens=5, pad_token_id=50256)
    result = out[0]['generated_text'][len(prompt):].strip().split('\n')[0]
    print(f"Run {i+1}: {result}")

# Challenger B (Temp 0.9)
print("\n=== Challenger B (Temp 0.9) ===")
for i in range(3):
    out = generator(prompt, do_sample=True, temperature=0.9, top_p=0.95, top_k=50, max_new_tokens=5, pad_token_id=50256)
    result = out[0]['generated_text'][len(prompt):].strip().split('\n')[0]
    print(f"Run {i+1}: {result}")
````

---

## Step 4 — Evaluate

Score the behaviour, not factual correctness.

|Configuration|Run|Output Summary|Score|Observation|
|---|---:|---|---:|---|
|Primary|1|Account Access|3/5|Consistent single-word label across all runs.|
|Primary|2|Account Access|3/5|Same as run 1.|
|Primary|3|Account Access|3/5|Same as run 1.|
|Challenger A|1|Account Access|3/5|Matched primary run.|
|Challenger A|2|Support|2/5|Output changed to generic 'Support'.|
|Challenger A|3|Account Access|3/5|Returned to Account Access.|
|Challenger B|1|Customer Care|1/5|Made up a category outside our prompt list.|
|Challenger B|2|Other|1/5|Output generic label 'Other'.|
|Challenger B|3|Electronic Payments|1/5|Made up another custom category label.|

### Questions

- **Which configuration won?** Primary (Greedy decoding, Temp = 0.0).
- **Why?** Because for classification we want the exact same output every single time. Temp = 0.0 stayed 100% consistent across all runs without making up random category names.
- **Did the evidence surprise you?** Not completely, but Challenger B (Temp 0.9) showed how higher temperature makes the model unpredictable, it started making up custom categories ("Customer Care", "Electronic Payments") that weren't even in our prompt list.

## Step 3 — Create One Representative Prompt

Choose ONE prompt that realistically represents this product.

```text
Write a short, professional email to a customer named Sarah whose package was delayed by 3 days because of bad weather. 
Apologize for the delay and give her a 15% discount code: WEATHER15.

Customer: Sarah
Reply:

````python
# ================================
# Run your experiments here
# ================================

import warnings
warnings.filterwarnings("ignore")

from transformers import pipeline

generator = pipeline("text-generation", model="gpt2")

# Few-shot prompt showing GPT-2 what a support email looks like
prompt = """Customer: Mark
Issue: Item broken in transit.
Reply: Dear Mark, We are so sorry your item arrived broken. Here is a 15% discount code: SORRY15 for your next order. Best, Support Team.

Customer: Sarah
Issue: Package delayed 3 days due to bad weather. Discount code: WEATHER15.
Reply: Dear Sarah,"""

# Primary (Temp 0.7)
print("=== Primary (Temp 0.7) ===")
for i in range(3):
    out = generator(prompt, do_sample=True, temperature=0.7, top_p=0.9, top_k=50, max_new_tokens=30, pad_token_id=50256)
    result = "Dear Sarah," + out[0]['generated_text'][len(prompt):].strip().split('\n')[0]
    print(f"Run {i+1}:\n{result}\n---")

# Challenger A (Greedy)
print("\n=== Challenger A (Greedy) ===")
for i in range(3):
    out = generator(prompt, do_sample=False, max_new_tokens=30, pad_token_id=50256)
    result = "Dear Sarah," + out[0]['generated_text'][len(prompt):].strip().split('\n')[0]
    print(f"Run {i+1}:\n{result}\n---")

# Challenger B (Temp 1.2)
print("\n=== Challenger B (Temp 1.2) ===")
for i in range(3):
    out = generator(prompt, do_sample=True, temperature=1.2, top_p=0.98, top_k=100, max_new_tokens=30, pad_token_id=50256)
    result = "Dear Sarah," + out[0]['generated_text'][len(prompt):].strip().split('\n')[0]
    print(f"Run {i+1}:\n{result}\n---")
````

## Step 4 — Evaluate

Score the behaviour, not factual correctness.

|Configuration|Run|Output Summary|Score|Observation|
|---|---:|---|---:|---|
|Primary|1|Late order notice|3/5|Polite tone, but stated order couldn't be processed.|
|Primary|2|Delayed 3 days apology|4/5|Good support response, mentioned 3-day weather delay.|
|Primary|3|Customer service notice|4/5|Good natural phrasing ('Thank you for your patience').|
|Challenger A|1|Repetitive apology|3/5|Correct info, but repeated 'delayed due to bad weather' twice.|
|Challenger A|2|Repetitive apology|3/5|Exact same output as run 1 (100% deterministic).|
|Challenger A|3|Repetitive apology|3/5|Exact same output as run 1.|
|Challenger B|1|Mail delivery issue|2/5|Confused story about mail delivery.|
|Challenger B|2|Store opening hours|2/5|Off-topic text about 9 a.m. store speed on July 11th.|
|Challenger B|3|200 dollar refund request|1/5|Completely off-topic (200 dollar refund request).|

### Questions

- **Which configuration won?** Primary (Temp = 0.7, Top-p = 0.9).
- **Why?** It gives nice, natural variations for customer service emails without getting robotic like Greedy or completely off-topic like High Temp.
- **Did the evidence surprise you?** Yes, Challenger B (Temp 1.2) got way too wild and started generating off-topic store hours and requesting 200 dollar refunds.

- **Primary** → Your recommendation

|Challenger B|0.0|N/A|1.0|0.0|Greedy|Testing greedy decoding to show how zero randomness gives boring and repetitive ideas.|

## Step 3 — Create One Representative Prompt

Choose ONE prompt that realistically represents this product.

```text
Product: HydroPure Smart Water Bottle
Campaign Angles:
1. Title: Pure Ocean. Target: Environmentalists. Slogan: Every sip cleans the sea.
2. Title: Smart Hydrate. Target: Athletes. Slogan: Track your flow.
3. Title:

````python
# ================================
# Run your experiments here
# ================================

import warnings
warnings.filterwarnings("ignore")

from transformers import pipeline

generator = pipeline("text-generation", model="gpt2")

# Prompt giving GPT-2 water bottle examples to follow
prompt = """Product: HydroPure Smart Water Bottle
Campaign Angles:
1. Title: Pure Ocean. Target: Environmentalists. Slogan: Every sip cleans the sea.
2. Title: Smart Hydrate. Target: Athletes. Slogan: Track your flow.
3. Title:"""

# Primary (Temp 0.95, Repetition Penalty 1.2)
print("=== Primary (Temp 0.95) ===")
for i in range(3):
    out = generator(prompt, do_sample=True, temperature=0.95, top_p=0.95, top_k=80, repetition_penalty=1.2, max_new_tokens=40, pad_token_id=50256)
    result = "3. Title:" + out[0]['generated_text'][len(prompt):].strip().split('\n')[0]
    print(f"Run {i+1}:\n{result}\n---")

# Challenger A (Temp 0.5)
print("\n=== Challenger A (Temp 0.5) ===")
for i in range(3):
    out = generator(prompt, do_sample=True, temperature=0.5, top_p=0.85, top_k=40, max_new_tokens=40, pad_token_id=50256)
    result = "3. Title:" + out[0]['generated_text'][len(prompt):].strip().split('\n')[0]
    print(f"Run {i+1}:\n{result}\n---")

# Challenger B (Greedy)
print("\n=== Challenger B (Greedy) ===")
for i in range(3):
    out = generator(prompt, do_sample=False, max_new_tokens=40, pad_token_id=50256)
    result = "3. Title:" + out[0]['generated_text'][len(prompt):].strip().split('\n')[0]
    print(f"Run {i+1}:\n{result}\n---")
````

### Questions

- **Which configuration won?** Challenger A (Temp = 0.5, Top-p = 0.85).
- **Why?** It produced clean, sensible marketing angles without falling into greedy repetition loops or high-temperature chaos.
- **Did the evidence surprise you?** Yes! Greedy decoding (Challenger B) produced the funny output "Target: Water. Slogan: Water." 3 times in a row, while High Temp (0.95) generated wild stories about Sydney bikes and Florida store towels.

## Final Recommendations

|Feature|Recommended Configuration|Reason|

---

## House Rule #1

> **Use Greedy decoding (Temp = 0.0) whenever the model output needs to strictly follow a format or go directly into automated code.**

---

## House Rule #2

> **Use moderate temperature (0.5 to 0.7) with sampling when writing or brainstorming to get creative variation while avoiding off-topic chaos.**
